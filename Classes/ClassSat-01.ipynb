```python
import altair as alt
from vega_datasets import data
import pandas as pd
```

```python
stocks = data.stocks()
stocks.head()
```

```python
stock_title = alt.TitleParams(
    "Google's stock experiencing heavier fluctuations than competitors",
    subtitle="Prices have been surging since 2009, but have still not reached"
)

stock_max_date = stocks[stocks['date'] == stocks['date'].max()]
print (stock_max_date)
stock_max_date. loc[stock_max_date['symbol'] == "IBM", 'price'] = 140
stock_max_date. loc[stock_max_date['symbol'] == "AMZN", 'price'] = 110

texts = alt. Chart(stock_max_date).mark_text(align='left', dx=2).encode(
    x='date',
    y='price',
    text='symbol',
    color=alt.Color('symbol', legend=None)
)

lines = alt. Chart(stocks, title=stock_title). mark_line(). encode(
    alt.X('date', title=None, axis=alt.Axis(tickCount=10)),
    alt.Y('price', title='Closing price', axis=alt.Axis(format='$s')),
    color = alt.Color('symbol'),
    tooltip=[
        'symbol',
        'date',
        'price',
        alt.Tooltip('price', format='$.2s')
    ]
).properties(
    width=500
)

lines + texts
```

```python
import pandas as pd
import altair as alt

dataset_url = 'https://raw.githubusercontent.com/erickedu85/dataset/master/penguins.csv'
df = pd.read_csv(dataset_url)
df
```

```python
df.info()
```

```python
print(f"Unique values in species: {df['species'].unique()}\n")
print(f"Unique values in island: {df['island'].unique()}")
```

```python
print("Unique values in Col1:", df['island'].unique())
```

```python
grafico = alt.Chart(df).mark_boxplot(
    size=40,          # Grosor de las cajas
    outliers=True     # Mostrar puntos atípicos
).encode(
    # Eje X: Especie (ocultamos el título del eje inferior para limpiar el diseño)
    x=alt.X('species:N', title=None, axis=alt.Axis(labels=False, ticks=False)),
    
    # Eje Y: Masa corporal. No empezamos en 0 para ver mejor las diferencias de distribución
    y=alt.Y('body_mass_g:Q', 
            title='Masa Corporal (gramos)', 
            scale=alt.Scale(zero=False)),
    
    # Color: Asignamos un color por especie para rápida identificación
    color=alt.Color('species:N', 
                    title='Especie', 
                    scale=alt.Scale(scheme='set2')),
    
    # Facetado (Small Multiples): Una columna por cada Isla
    column=alt.Column('island:N', 
                      title='Isla', 
                      header=alt.Header(labelOrient='bottom', titleOrient='bottom', labelFontSize=12))
    
).properties(
    width=150,  # Ancho de cada subgráfico
    height=350, # Alto de los gráficos
    title="Distribución de Masa Corporal de Pingüinos por Especie e Isla"
).configure_title(
    fontSize=18,
    anchor='start',
    offset=20
).configure_facet(
    spacing=15 # Espacio entre los paneles
).configure_view(
    stroke=None # Quitar el borde exterior de los paneles para un look más moderno
)
grafico.show()
```
