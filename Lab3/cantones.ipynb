```python
def normalize_text(text):
    if pd.isna(text):
        return text
    
    # Convert to uppercase and strip whitespaces
    text = str(text).upper().strip()
    
    # Remove accents/diacritics (e.g., 'MANABÍ' -> 'MANABI')
    text = ''.join(c for c in unicodedata.normalize('NFD', text)
                   if unicodedata.category(c) != 'Mn')
    return text
```

```python
def main():
    print("Loading datasets...")
    # 1. Load the datasets
    # Adjust the filepaths if your files are located in a different directory
    df = pd.read_csv('cantones_limpio.csv')
    gdf = gpd.read_file('cantones.geojson')

    print("Normalizing geographic names...")
    # 2. Normalize the matching columns (Canton names)
    df['canton_norm'] = df['canton'].apply(normalize_text)
    gdf['DPA_DESCAN_norm'] = gdf['DPA_DESCAN'].apply(normalize_text)

    print("Aggregating insights...")
    # 3. Create Insights (Aggregations)
    
    # Insight A: Total deaths per canton
    total_deaths = df.groupby('canton_norm').size().reset_index(name='Total_Muertes')

    # Insight B: Deaths broken down by sex
    sex_deaths = pd.crosstab(df['canton_norm'], df['sexo']).reset_index()
    # Rename columns to avoid spaces/weird characters if any exist
    sex_deaths.columns.name = None 

    # Insight C: Deaths broken down by 'tipo_muerte' (Type of death)
    tipo_deaths = pd.crosstab(df['canton_norm'], df['tipo_muerte']).reset_index()
    tipo_deaths.columns.name = None 

    # 4. Merge all insights into a single DataFrame
    agg_df = total_deaths.merge(sex_deaths, on='canton_norm', how='left')
    agg_df = agg_df.merge(tipo_deaths, on='canton_norm', how='left')

    print("Merging insights with geographic boundaries...")
    # 5. Merge the aggregations with the GeoDataFrame
    merged_gdf = gdf.merge(agg_df, left_on='DPA_DESCAN_norm', right_on='canton_norm', how='left')

    # 6. Clean up the final dataset
    # Drop the temporary normalization columns
    merged_gdf = merged_gdf.drop(columns=['DPA_DESCAN_norm', 'canton_norm'])
    
    # Fill NaN values with 0 for cantons that had no reported crimes in the CSV
    metric_cols = agg_df.columns.drop('canton_norm')
    merged_gdf[metric_cols] = merged_gdf[metric_cols].fillna(0)

    print("Exporting Kepler.gl ready GeoJSON...")
    # 7. Save the final GeoDataFrame to a new GeoJSON file
    output_filename = 'kepler_ecuador_crimes.geojson'
    merged_gdf.to_file(output_filename, driver='GeoJSON')
    print(f"Success! File saved as: {output_filename}")
```

```python
main()
```
